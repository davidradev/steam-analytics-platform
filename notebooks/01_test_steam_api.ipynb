```python
import os
import requests
import pandas as pd
from dotenv import load_dotenv

# Cargar variables desde el archivo .env
load_dotenv(dotenv_path="../.env")
api_key = os.getenv("STEAM_API_KEY")
```

```python
url = f"https://api.steampowered.com/ISteamChartsService/GetGamesByConcurrentPlayers/v1/?key={api_key}"
```

```python
response = requests.get(url)

if response.status_code == 200:
    data = response.json()
    # Navegamos en el JSON hasta la lista de juegos
    games_list = data['response']['ranks']
    df = pd.DataFrame(games_list)
    print("¡Conexión exitosa!")
    print(df.head()) # Ver los top 5 juegos
else:
    print(f"Error: {response.status_code}")
```
